# Ejercicio

Una cepa de E. coli tiene una tasa de crecimiento tal que un individuo se duplica cada 30 min.

Calcula en cuanto tiempo esta cepa

### a) Llenaria el volumen de la tierra (considerandola como una esfera hueca)

```python
import math as mt
```

```python
rt = 6371
```

```python
rec= 0.5
```

```python
rtm = rt * 1e3
```

```python
recm = rec * 1e-6
```

```python
N = (rtm/recm) ** 3
```

```python
print (N)
```

```python
t = mt.log2(N)
```

```python
td = t/48
```

```python
td
```

# b) cubriria la superfice de la tierra

```python
import math as mt
```

```python
rt = 6371
```

```python
rec= 0.5
```

```python
rtm = rt * 1e3
recm = rec * 1e-6
```

```python
Ns = (rtm/recm) ** 2
```

```python
Ns
```

```python
ts = mt.log2(Ns)
```

```python
tds = ts/48
```

```python
tds
```
